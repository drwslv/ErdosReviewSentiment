## Bag-of-words text pre-processing

```python
import numpy as np
import pandas as pd

df = pd.read_csv('./data/wyoming.csv')
df.shape
```

```python
df.loc[0,['text']]
```

`CountVectorizer()` performs:

* Lowercasing

* Removes punctuation

* Removes stopwords (note: may want to use custom stopword dictionary)

* Tokenizes

* Converts to sparse matrix format (binary encoded vector relating to implicit vocabulary dictionary, unless specified)

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer(
    strip_accents='unicode',
    lowercase=True,
    stop_words='english', max_features=5000
)
```

```python
X = vectorizer.fit_transform(df['text'])
```

### Verify that encoding aligns with original text

```python
features = vectorizer.get_feature_names_out()
features
```

```python
X.shape # (observations, dictionary size)
```

```python
X[0,:].nonzero()[1]
```

```python
nonzero_indices = X[0,:].nonzero()[1]

for idx in nonzero_indices:
    print(f'Word: {features[idx]}, Count: {X[0, idx]}')
```

```python
y = df['rating'] - 1
```

### Train-test split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```
