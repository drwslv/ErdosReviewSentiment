## Convert raw json to csv

```python
import json
import pandas as pd
```

### Read-in 10-core file

```python
df_core = pd.read_json('./data/review-Wyoming_10.json.gz', compression='gzip', lines=True, dtype={'text': 'str'})
```

```python
df_core.columns
```

```python
df_core = df_core[['gmap_id','rating','time','text']]
```

```python
df_core.head(2)
```

```python
df_core = df_core[df_core['text']!='None']
```

```python
df_core.info()
```

### Read-in meta file

```python
df_meta = pd.read_json('./data/meta-Wyoming.json.gz', compression='gzip', lines=True)
```

```python
df_meta.columns
```

```python
df_meta = df_meta[['gmap_id','category','latitude','longitude']]
df_meta = df_meta.loc[ ~df_meta['category'].isna() ]
df_meta = df_meta.drop_duplicates(subset='gmap_id')
```

```python
df_meta.head(2)
```

```python
df_meta.info()
```

### Merge

```python
df = pd.merge(df_core, df_meta, how='left', on='gmap_id')
```

```python
df.info()
```

```python
df
```

```python
df.to_csv('./data/wyoming.csv', index=False)
```
